```python
import os

import pandas as pd
from dotenv import load_dotenv
from database import engine

load_dotenv(override=True)

# Read CSV
df = pd.read_csv("../data/grocery_sales.csv")

# Remove unnecessary index column from CSV
df = df.drop(columns=["index"])

# Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Upload to PostgreSQL
df.to_sql(
    name="grocery_sales",
    con=engine,
    schema="sales",
    if_exists="replace",
    index=False
)

print(f"{len(df)} rows uploaded!")
```
