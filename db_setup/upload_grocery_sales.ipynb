```python
import pandas as pd
from db_connect import get_engine

load_dotenv(override=True)

# Read CSV
df = pd.read_csv("../data/grocery_sales.csv")

# Convert Date
df["Date"] = pd.to_datetime(df["Date"])

# Upload to PostgreSQL
df.to_sql(
    name="grocery_sales",
    con=get_engine(),
    schema="sales",
    if_exists="replace",
    index=False
)

print(f"{len(df)} rows uploaded!")
```
