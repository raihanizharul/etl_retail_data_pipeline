```python
import pandas as pd
import os
from db_setup.db_connect import get_engine
```

```python
# Read grocery_sales from database
grocery_sales = pd.read_sql('''
                       SELECT *
                       FROM sales.grocery_sales     
                ''' , con=get_engine())

# Print head grocery_sales
print(grocery_sales.head())
```

```python
# Create the transform() function with one parameter: "raw_data"
def transform(raw_data):
    # Fill Date values with ffill
    raw_data["Date"] = raw_data["Date"].ffill()
    
    # Fill Weekly_Sales null values with mean
    raw_data["Weekly_Sales"] = raw_data["Weekly_Sales"].fillna(raw_data["Weekly_Sales"].mean())
    # Fill CPI null values with mean
    raw_data["CPI"] = raw_data["CPI"].fillna(raw_data["CPI"].mean())
    # FIll Unemployment null values with mean
    raw_data["Unemployment"] = raw_data["Unemployment"].fillna(raw_data["Unemployment"].mean())
    
    # Keep rows where weekly sales are over $10,000
    raw_data = raw_data.loc[raw_data["Weekly_Sales"] > 10000, :]
    
    # Add Month column
    raw_data["Month"] = raw_data["Date"].dt.month
    
    # Use necessary column
    raw_data = raw_data.loc[:, ["Store_ID", "Month", "Dept", "IsHoliday", "Weekly_Sales", "CPI", "Unemployment"]]

    return raw_data
```

```python
# Call the transform() function and pass the merged DataFrame
clean_data = transform(merged_df)

# Print clean_data
print(clean_data.head())
```

```python
# Create the avg_weekly_sales_per_month function that takes in the cleaned data from the last step
def avg_weekly_sales_per_month(clean_data):
    return (clean_data.groupby("Month")
            .agg({"Weekly_Sales": 'mean'})
            .round(2)
            .reset_index()
            .rename(columns={"Weekly_Sales": "Avg_Sales"})
    )
```

```python
# Call the avg_weekly_sales_per_month() function and pass the cleaned DataFrame
agg_data = avg_weekly_sales_per_month(clean_data)

# Print agg_data
print(agg_data)
```

```python
# Create the load() function that takes in the cleaned DataFrame and the aggregated one with the paths where they are going to be stored
def load(full_data, full_data_file_path, agg_data, agg_data_file_path):
    full_data.to_csv(full_data_file_path, index=False)
    agg_data.to_csv(agg_data_file_path, index=False)
```

```python
# Call the load() function and pass the cleaned and aggregated DataFrames with their paths    
clean_data_path = 'data/clean_data.csv'
agg_data_path = 'data/agg_data.csv'

load(clean_data, clean_data_path, agg_data, agg_data_path)
```

```python
# Create the validation() function with one parameter: file_path - to check whether the previous function was correctly executed
def validation(file_path):
    exists = True
    for path in file_path:
        if os.path.exists(path) == True:
            exists = True
        else:
            exists = False
    return exists
```

```python
# Call the validation() function and pass first, the cleaned DataFrame path, and then the aggregated DataFrame path
print(validation([clean_data_path, agg_data_path]))
```
